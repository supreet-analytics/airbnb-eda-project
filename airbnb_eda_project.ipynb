# Data Cleaning

```python
# changing the format of the column: last review

df['last_review']=pd.to_datetime(df['last_review'],errors='coerce')
df.info()
```

```python
# filling the missing values for the columns: host identify verified, last review, reviews per month, review rate number

df.fillna({'last_review':df['last_review'].min(),'number_of_reviews':0,'reviews_per_month':0,'number_of_reviews_ltm':0},inplace=True)
df.isnull().sum()
```

```python
# removing column: house_rules and license as these columns have a lot of null values

df.drop(columns=['license'],inplace=True)
```

```python
# removing rows with null values: Name, Host_Name, Price

df.dropna(inplace=True)
df.isnull().sum()
```

```python
# replacing $ sign from: price and service_fee

df['price']=df['price'].replace(r'[\$,]','',regex=True).astype(float)
```

```python
# adding a column filled_rating as this will help in understanding the relationship between rating and neighbourhood_group

df[df['rating'].str.contains('No rating')]

df['filled_rating']=df['rating']
df['filled_rating']=pd.to_numeric(df['filled_rating'],errors='coerce')
df['filled_rating']=df['filled_rating'].fillna(df['filled_rating'].mean())
df['filled_rating']=df['filled_rating'].astype(float)
df
```

```python
df.info()
```

```python
# identifying the number of duplicate values

df.duplicated().sum()
df[df.duplicated()]
```

```python
# removing duplicate values

df.drop_duplicates(inplace=True)
df.info()
```

# Visualization


**Univariate Analysis**

```python
df['price']
```

```python
# identifying outliers using boxplot

plt.figure(figsize=(10,6))
sns.boxplot(df['price'],color='skyblue')
plt.title('Identifying Outliers in Price')
plt.ylabel('Price $')
plt.show()
```

```python
# identifying outliers less than 1500

data=df[df['price']<1500]
plt.figure(figsize=(10,6))
sns.boxplot(data['price'],color='lightseagreen')
plt.ylabel('Price $')
plt.show()
```

```python
# What is the distribution of listing prices?


plt.figure(figsize=(12,7))
sns.histplot(data['price'],color='orchid',bins=30,kde=True)
plt.title('Distribution of Airbnb Prices')
plt.xlabel('Price $')
plt.ylabel('Frequency')
plt.show()
```

```python
room_type_mean=df.groupby('room_type').mean(numeric_only=True)
room_type_mean
```

```python
# What kind of room types are the most common?

explodes=[0,0.1,0,0]
plt.figure(figsize=(12,7))
plt.pie(room_type_mean['price'],labels=df['room_type'].unique(),explode=explodes,autopct='%.2f%%',pctdistance=0.7,startangle=90)
plt.title('Most Common Room Types')
plt.show()
```

**Feature Engineering**

```python
# what is the price of each bed?

df['price_per_bed']=df['price']/df['beds']
df.head()

# manhattan has the highest price for beds but does that depend on the type of room available?

df.groupby('neighbourhood_group')['price_per_bed'].mean()
df.groupby(['neighbourhood_group','room_type'])['price_per_bed'].mean()
df.groupby(['neighbourhood_group','room_type'])['price_per_bed'].count()
```

**Bivariate Analysis**

```python
# how are listings distributed across different neighbourhood groups?

df.nunique()
plt.figure(figsize=(10,6))
sns.countplot(y='neighbourhood_group',data=df,hue='neighbourhood_group',palette='Accent',order=df['neighbourhood_group'].value_counts().index)
plt.title('Number of Listings by Neighbourhood Group')
plt.xlabel('Count')
plt.ylabel('Neighbourhood Group')
plt.show()
```

```python
# is there a relationship between neighbourhood group and availabilty 365?

plt.figure(figsize=(10,6))
sns.violinplot(x='neighbourhood_group', y='availability_365', data=df,hue='neighbourhood_group',palette='Pastel2')
plt.title("Distribution of Availability across Neighbourhood Groups")
plt.xticks(rotation=45)
plt.xlabel('Neighbourhood Groups')
plt.ylabel('365 Availability')
plt.show()
```

```python
# relationship between neighbouhood group and availability 365 via barplot

plt.figure(figsize=(8,5))
sns.barplot(x='neighbourhood_group', y='availability_365', data=df,hue='neighbourhood_group',palette='summer', estimator=np.mean)
plt.title("Average Availability by Neighbourhood Group")
plt.xlabel('Neighbourhood Group')
plt.ylabel('365 Availability')
plt.xticks(rotation=45)
plt.show()
```

```python
# what is the relationship between price and room type?

plt.figure(figsize=(10,6))
sns.boxplot(x='room_type',y='availability_365',data=df,hue='room_type',palette='Accent')
plt.title('365 Availability Vs Room Type')
plt.xlabel('Room Type')
plt.ylabel('365 Availability')
plt.show()

# to visualize boxplot more clearily price > 500 were excluded temporarily
```

```python
# if the listings in Bronx are available 365 and the prices are reasonable too then why are the listings in Bronx too low?

plt.figure(figsize=(8,5))
sns.barplot(x='neighbourhood_group', y='price', data=df,hue='room_type',palette='Set2',estimator=np.mean)
plt.title("Average Price by Neighbourhood Group")
plt.xlabel('Neighbourhood Group')
plt.ylabel('Price $')
plt.xticks(rotation=45)
plt.show()
```

```python
# does more price means more number of reviews?

plt.figure(figsize=(12,6))
price_review=df[df['price']<=1500]
sns.scatterplot(x='number_of_reviews',y='price',data=price_review,hue='neighbourhood_group',palette='Set1')
plt.title('Locality and Review Dependency')
plt.xlabel('Number of Reviews')
plt.ylabel('Price $')
plt.legend(title='Neighbourhood Group')
plt.show()
```

```python
# which neighbourhood group has more rating?

clean_rating=df[df['filled_rating']<=5]
sns.barplot(x='neighbourhood_group',y='filled_rating',data=clean_rating,hue='neighbourhood_group',palette='Pastel1')
plt.title('Ratings in each Neighbourhood Group')
plt.xlabel('Neighbourhood Group')
plt.ylabel('Filled Rating')
plt.show()
```

```python
# how has the number of reviews changed with time?

reviews_over_time=df.groupby(df['last_review'].dt.to_period('M')).size()
plt.figure(figsize=(12,6))
reviews_over_time.plot(title='Number of Reviews Over Time',xlabel='Date',ylabel='Number of Reviews',colormap='summer')
plt.show()
```
